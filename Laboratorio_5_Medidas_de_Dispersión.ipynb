3. Calcula el rango de la variable *Pago base*, ¿este valor representa la realidad de todos los trabajadores?

```python
max_pb = df_1['Pago base'].max()
min_pb = df_1['Pago base'].min()
rango = max_pb - min_pb
print(f'El rango de los sueldos de los trabajores es de ${rango}[USD]')
```

Este valor indica que existe una gran diferencia eltre el sueldo mas alto y el sueldo mas bajo

4. Calcula la media, mediana, desviación estándar y coeficiente de variación del sueldo base. ¿La dispersión es alta o baja respecto al promedio? ¿Qué medida (media o mediana) consideras más representativa para describir el sueldo típico? Justifica tu respuesta.



```python
media_pb = df_1['Pago base'].mean()
mediana_pb = df_1['Pago base'].median()
desvest_pb = df_1['Pago base'].std()

cv_pb = (desvest_pb/media_pb)*100

print(f'La media de los sueldos es de ${media_pb:.0f}[USD]')
print(f'La mediana de los sueldos es de ${mediana_pb:.0f}[USD]')
print(f'La desviacion estandar de los sueldos es de ${desvest_pb:.0f}[USD]')
print(f'El coeficiente de variacion de los sueldos es de {cv_pb:.1f}%')


```

Dado los resultados diremos que los sueldos presentan una alta dispersion, por lo mismo, consideraremos la mediana como un valor mas representativo.
